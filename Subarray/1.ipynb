# Find the subarray with the maximum sum (Brute Force).


```python
arr = [1,2,3]

max_sum = float('-inf')

for i in range(len(arr)):
    for j in range(i,len(arr)):

        subarray = arr[i:j+1]
        print(subarray)

        sum_subarray = sum(subarray)
        print(sum_subarray)

        if sum_subarray > max_sum:
            max_sum = sum_subarray

print("Maximum Subarray Sum:", max_sum)

        
```

# Find the subarray having a given sum K (positive numbers)

```python
arr = list(map(int,input("Enter the array :").split()))
print(f"Array : {arr}")

K = int(input("Enter the sum"))
print(f"Input sum : {K}")

for i in range(len(arr)):
    for j in range(i,len(arr)):
        substring = arr[i:j+1]   # substring
        
        sum_substring = sum(substring)  # sum of substring
        

        if K == sum_substring:  # comparision
              
            print(f"{substring} having given sum {K}")
        
```

# Count subarrays whose sum equals K (Brute Force).

```python
arr = list(map(int,input("Enter the value").split()))
print(f"Array : {arr}")

K = int(input("Enter the value"))
print(f"Value : {K}")

count = 0 

for i in range(len(arr)):
    for j in range(i,len(arr)):
        subarray = arr[i:j+1]

        sum_subarray = sum(subarray)

        if sum_subarray == K:
            count += 1
            
print("Total subarrays:", count)
```

# Find the longest subarray having sum K (positive numbers).

```python
arr = list(map(int,input("Enter the value").split()))
print(f"Array : {arr}")

K = int(input("Enter the value"))
print(f"Sum Value : {K}")

max_length = 0
longest_subarray = []

for i in range(len(arr)):
    for j in range(i,len(arr)):
        subarray = arr[i:j+1]

        sum_subarray = sum(subarray)

        len_subarray = len(subarray)
        
        if sum_subarray == K:

            if len_subarray > max_length:
                max_length = len_subarray
                longest_subarray = subarray

print("Longest Subarray:", longest_subarray)
print("Maximum Length:", max_length)
        
        
```

# Find the subarray with the maximum average.

```python
arr = list(map(int,input("Enter the value").split()))
print(f"Array : {arr}")

max_subarray = float('-inf')
for i in range(len(arr)):
    for j in range(i,len(arr)):

        subarray = arr[i:j+1]
        print(subarray)
        sum_subarray = sum(subarray)

        len_subarray = len(subarray)

        avg_subarray = sum_subarray/len_subarray
        print(avg_subarray)

        if avg_subarray > max_subarray:
            max_subarray = avg_subarray

print(f"Maximum subarray with max_avg : {max_subarray}") 

        
```

# Find the maximum sum of a subarray of size K.

- too isme sbse phle hume K size ka subarray nikalna h... than complare krna h ki kis subarray kaa size maximum h .. aur usko hi output me dena h ...

```python
arr = list(map(int, input("Enter the array values: ").split()))
print(f"Array: {arr}")

K = int(input("Enter K: "))

max_sum = float('-inf')
max_subarray = []

for i in range(len(arr)-K+1):

    subarray = arr[i:i+K]

    sum_subarray = sum(subarray)
    len_subarray = len(subarray)

    if len_subarray == K and sum_subarray > max_sum:
        max_sum = sum_subarray
        max_subarray = subarray

print("Maximum Subarray:", max_subarray)
print("Maximum Sum:", max_sum)
```
