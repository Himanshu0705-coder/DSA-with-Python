# Largest Common Prefix


```python
strs = ["flower", "flow", "flight"]
print("string elements::", strs)
prefix = strs[0]

for word in strs:
    while not word.startswith(prefix):
        prefix = prefix[:-1]

print("Longest Common Prefix:","*",prefix,"*")
```

```python
words = input("Enter the words:").split()
print("Enter the words:: ",words)

prefix = words[0]

for word in words:
    while not word.startswith(prefix):
        prefix = prefix[:-1]
    
    if prefix == "":
        print("NO similar prefix present")

print("Prefix values: ",prefix)
```
