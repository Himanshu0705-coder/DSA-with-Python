# Container with more water

```python
height = [1, 8, 6, 2, 5, 4, 8, 3, 7]

left = 0
right = len(height) - 1
max_water = 0

while left < right:

    widht = right - left
    h  = min(height[left],height[right])
    area = widht * h

    max_water = max(max_water, area)

    if height[left] < height[right]:
        left += 1

    else:
        right -= 1
        
print("So the maximum water contained :- ",max_water)
```
