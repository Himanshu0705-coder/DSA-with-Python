Write a code to get the sum of two numbers equals to the target.

```python
num = [2,4,5,3,6,7,4]
target = 9

for i in range(len(num)):
    for j in range(i+1,len(num)):
        if num[i] + num[j] == target:
            print(i,j)
    
```
