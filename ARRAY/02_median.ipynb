# Median of an array 

# 1. Odd array
 

```python
num = [1,2,6,4,3,4]
num.sort()

median = len(num) // 2

print("Sorted array :- ",num)
print("Median :- ", median)
```

```python
num = list(map(int,input("Enter the values").split()))
print("values :- " , num)

num.sort()
n = len(num)
print("sorted array :-", num)

median = num[n//2]
    
print("median:-",median)
```

# Even array

```python
num = list(map(int, input("Enter array elements: ").split()))
print("Unsorted array: ",num)
num.sort()
n = len(num)

if n % 2 == 1:
    median = num[n // 2]
else:
    median = (num[n // 2 - 1] + num[n // 2]) / 2

print("Sorted array:", num)
print("Median:", median)
```

# median of two sorted array

```python
num1 = [1,3,2]
num2 = [4,5,6]
merge = num1 + num2
print("Merged array:- ",merge)
arr = sorted(merge)
print("Sorted array:- ",arr)

n = len(arr)

if n % 2 == 1:
    median = arr[n // 2]
    print("odd median:",median)
else:
    median = (arr[n // 2] + arr[n//2 +1]) / 2
    print("even median",median)

```
