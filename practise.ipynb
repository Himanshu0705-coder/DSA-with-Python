# Find the Second-Largest Element

```python
# Find the Second-Largest Element

def second_largest(arr):
    largest = float("-inf")
    second =float("-inf")

    for num in arr:
        if num > largest:
            second = largest
            largest = num

        elif num > second and num != largest:
            second = num

    if second == float("-inf"):
        return -1

    return second

arr = list(map(int,input("Enter the list elements").split()))
print(f"arr_list : {arr}")
print(f"sported_arr : {sorted(arr)}")
print(f"second highest : {second_largest(arr)}")
```

# Find duplicate element

```python
def duplicate_arr(arr):
    arr.sort()

    left = 0 
    right = 1

    while left < len(arr):
        if arr[left] == arr[right]:
            return arr[left]

        else:
            left += 1
            right -= 1
    return -1

arr = [1,2,3,4,5,5,6,6]
print(f"duplicate_arr : {duplicate_arr(arr)}")
    
```

# Two Sum

```python
def two_sum(arr, target):
    left = 0
    right = len(arr) - 1

    while left < right:
        total = arr[left] + arr[right]

        if total == target:
            return [arr[left], arr[right]]

        elif total < target:
            left += 1

        else:
            right -= 1

    return []


arr = [2, 7, 11, 15]
print(two_sum(arr, 9))
```

# Move all zeros to the end

```python
def zero(arr):
    left = 0 

    for right in range(len(arr)):
        if arr[right] != 0:
            arr[left], arr[right] = arr[right], arr[left]
            left += 1

    return arr

arr = [0,1,0,3,12]
print(zero(arr))
```

# Find the missing number

```python
def missing_number(arr):
    n = len(arr)

    expected = n * (n + 1) // 2
    actual = sum(arr)

    return expected - actual


arr = [13, 0, 1]
print(missing_number(arr))
```

# Find the majority element
- Find the element appearing more than n/2 times.

```python
def majority_element(arr):
    candidate = None
    count = 0

    for num in arr:
        if count == 0:
            candidate = num

        elif num == candidate:
            count += 1
        else:
            count -= 1

    return candidate


arr = [2, 2, 1, 1, 1, 2,2]
print(majority_element(arr))
```

# Maximum subarray sum (Kadane's Algorithm)

```python
def max_subarray_sum(arr):
    current_sum = arr[0]
    max_sum = arr[0]

    for num in arr[1:]:

        current_sum = max(num, current_sum + num)
        
        max_sum = max(max_sum, current_sum)
        

    return max_sum


arr = [-2, 1, -3, 4, -1, 2, 1, -5, 4]

print(max_subarray_sum(arr))
```

# Find duplicate number

```python
def find_duplicate(nums):

    slow = nums[0]
    fast = nums[0]

    # Phase 1: Find intersection
    while True:
        slow = nums[slow]
        fast = nums[nums[fast]]

        if slow == fast:
            break

    # Phase 2: Find entrance of cycle
    slow = nums[0]

    while slow != fast:
        slow = nums[slow]
        fast = nums[fast]

    return slow


arr = [1, 3, 4, 2, 2]
print(find_duplicate(arr))
```

# Merge two sorted arrays

```python
def merge_sorted(arr1, arr2):

    left = 0
    right = 0

    result = []

    while left < len(arr1) and right < len(arr2):

        if arr1[left] < arr2[right]:
            result.append(arr1[left])
            left += 1

        else:
            result.append(arr2[right])
            right += 1

    result.extend(arr1[left:])
    result.extend(arr2[right:])

    return result


arr1 = [1, 3, 5]
arr2 = [2, 4, 6]

print(merge_sorted(arr1, arr2))
```

# Container With Most Water

```python
height = [1, 8, 6, 2, 5, 4, 8, 3, 7]

left = 0
right = len(height) - 1
max_area = 0

while left < right:
    width = right - left
    area = width * min(height[left], height[right])

    max_area = max(max_area, area)

    if height[left] < height[right]:
        left += 1
    else:
        right -= 1

print(max_area)
```

# Maximum sum subarray of size K

```python
arr = [2, 1, 5, 1, 3, 2]
k = 3

window_sum = sum(arr[:k])
max_sum = window_sum

for i in range(k, len(arr)):
    window_sum += arr[i]
    window_sum -= arr[i-k]

    max_sum = max(max_sum, window_sum)

print(max_sum)
```

# Linked List

# Reverse a linked list

```python
def reverseList(head):
    prev = None
    curr = head

    while curr:
        next_node = curr.next
        curr.next = prev
        prev = curr
        curr = next_node

    return prev
```

# Find the middle of a linked list

```python
def middleNode(head):
    slow = head
    fast = head

    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next

    return slow
```

# Detect a cycle in a linked list

```python
def hasCycle(head):
    slow = head
    fast = head

    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next

        if slow == fast:
            return True

    return False
```

# Find the nth node from the end

```python
def nthFromEnd(head, n):
    slow = head
    fast = head

    for _ in range(n):
        if fast is None:
            return None
        fast = fast.next

    while fast:
        slow = slow.next
        fast = fast.next

    return slow
```

```python
def detectCycle(head):
    slow = head
    fast = head

    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next

        if slow == fast:
            slow = head

            while slow != fast:
                slow = slow.next
                fast = fast.next

            return slow

    return None
```
