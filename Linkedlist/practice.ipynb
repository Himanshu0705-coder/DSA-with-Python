# Linkedlist
A Linked List is a linear data structure in which elements are stored in separate objects called Nodes.

- Head
- Data
- Node
- Null

```python
class Node:
    def __init__(self,data):
        self.data= data
        self.next = None
        
```

```python
node1 = Node(10)

print(node1.data)
print(node1.next)
```

```python
node1 = Node(20)
node2 = Node(30)
node3 = Node(40)
node4 = Node(50)

node1.next = node2
node2.next = node3
node3.next = node4

head = node1
```

```python
print(head.data)
print(head.next.data)
print(head.next.next.data)
print(head.next.next.next.data)
print(head.next.next.next.next)
```

# How to traverse a Linked List?
Traversal means visiting every node one by one.

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


node1 = Node(10)
node2 = Node(20)
node3 = Node(30)

node1.next = node2
node2.next = node3

head = node1


temp = head

while temp is not None:
    print(temp.data)
    temp = temp.next
```

# Insert at beginning

```python
new_node = Node(5)

new_node.next = head
head = new_node
```

```python
print(head.data)     # or print(new_node.data)
```

# Insert at end

```python
new_node1 = Node(40)

temp = head

while temp.next is not None:
    temp = temp.next

temp.next = new_node1
```

```python
print(new_node1.data)
```

```python
head = head.next
print(head.data)
```

```python
prev = None
curr = head

while curr is not None:

    next_node = curr.next

    curr.next = prev          # Previous node

    prev = curr              # Current node

    curr = next_node          # next_node: Saves the next node before changing the link

head = prev
```
