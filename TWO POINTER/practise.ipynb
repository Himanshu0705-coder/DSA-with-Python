# Reverse an array

```python
arr = [10,20,30,40]

left = 0
right = len(arr) - 1

while left < right:
    arr[left],arr[right] = arr[right],arr[left]
    left += 1
    right -= 1
print(arr)
```

# Remove duplicates from a sorted array.

```python
arr = [10,10,20,30]

left = 0
right = len(arr)-1

while left < len(arr)-1:
    if arr[left] == arr[left+1]:
        arr.pop(left+1)
    else:
        left+=1
        

print(arr)

```

# Find two numbers whose sum equals the target.

```python
arr = [10,20,30,40]
target = 70

left = 0
right = len(arr)-1

while left < right :
    total = arr[left] + arr[right]

    if total == target :
        print(f"index : {left,right}")
        print(f"element : {arr[left],arr[right]}")
        break
    elif total < target :
        left +=1
    else :
        right -= 1
```

# Check Palindrom



```python
string = input("Enter your input")
print(f"String : {string}")

left = 0 
right = len(string) - 1

while left < len(string) - 1:
    if string[left] != string[right]:
        print("Not Palidrom ")
        break
    left +=1
    right -=1
else :
        
    print("palidrom")
```

```python
s = "madam"

left = 0
right = len(s) - 1

while left < right:

    if s[left] != s[right]:
        print("Not Palindrome")
        break

    left += 1
    right -= 1

else:
    print("Palindrome")
```

```python
s = "A man, a plan, a canal: Panama"

left = 0
right = len(s) - 1

while left < right:

    if not s[left].isalnum():
        left += 1

    elif not s[right].isalnum():
        right -= 1

    elif s[left].lower() != s[right].lower():
        print(False)
        break

    else:
        left += 1
        right -= 1

else:
    print(True)
```
