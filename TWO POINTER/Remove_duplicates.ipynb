# REMOVE DUPLICATE FROM A SORTED ARRAY


```python
num = [1,2,3,4,1,2,3,5,4,3]

unique = []

for i in num:
    if i not in unique:
        unique.append(i)
print(unique)
```

```python
numbers = list(map(int,input("Enter the values").split()))
print("Unsorted numbers:" ,numbers)

numbers.sort()

unique = []   # container to store the numbers

for i in numbers:
    if i not in unique:
        unique.append(i)
print("Unique numbers: ",unique) 
```
