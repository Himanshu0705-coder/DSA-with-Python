# Rearrange words

Given an array arr[] consisting of only 0's and 1's. Modify the array in-place to segregate 0s onto the left side and 1s onto the right side of the array.

```python
number = [1,0,1,0,0,0,0,1,1,1,0,1]
number.sort()
print(number)
```

```python
sorted(number)
print(number)
```

```python
number = [1,0,1,0,0,0,0,1,1,1,0,1]

left = 0
right = len(number) - 1

while left < right:
    if number[left] == 0:
        left += 1
    elif number[right] == 1:
        right -= 1
    
    else:
        number[left],number[right] = number[right],number[left]

print(number)
    
```
